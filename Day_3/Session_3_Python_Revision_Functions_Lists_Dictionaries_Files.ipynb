## 3. Dictionaries
Dictionaries store key-value pairs.

```python
student={
    "name":"Arun",
    "coding":85,
    "attendance":92
}
print(student["name"])
student["aptitude"]=78

for k,v in student.items():
    print(k,v)

students=[
 {"name":"Arun","score":85},
 {"name":"Priya","score":92},
 {"name":"Kavin","score":45}
]

for s in students:
    print(s["name"], s["score"])
```

## 6. Reusable Code Practice

```python
students=[
 {"name":"Arun","coding":85,"aptitude":78},
 {"name":"Priya","coding":92,"aptitude":88},
 {"name":"Kavin","coding":45,"aptitude":52}
]

def average(student):
    return (student["coding"]+student["aptitude"])/2

def status(avg):
    return "Ready" if avg>=70 else "Needs Improvement"

for s in students:
    avg=average(s)
    print(s["name"], avg, status(avg))
```

## 7. Mini Project - Student Result Management

```python
students=[
 {"name":"Arun","marks":[80,75,90]},
 {"name":"Priya","marks":[92,88,91]},
 {"name":"Kavin","marks":[45,52,60]}
]

def avg(marks):
    return sum(marks)/len(marks)

with open("results.txt","w") as file:
    for s in students:
        a=avg(s["marks"])
        result="Pass" if a>=50 else "Fail"
        line=f"{s['name']} | {a:.2f} | {result}"
        print(line)
        file.write(line+"\n")
```
