## 1. Loading the Dataset

`sklearn.datasets.load_wine()` returns feature data and target labels. We convert it into a single pandas DataFrame to work with.


```python
wine = load_wine()

df = pd.DataFrame(wine.data, columns=wine.feature_names)
df['target'] = wine.target
df['wine_class'] = df['target'].map(dict(enumerate(wine.target_names)))

df.to_csv('wine.csv', index=False)
df.head()
```

**About the dataset:** each row is a wine sample. Columns like `alcohol`, `malic_acid`, `color_intensity`, `proline`, etc. are chemical measurements. `wine_class` tells us which of the three cultivars (`class_0`, `class_1`, `class_2`) the sample belongs to — this will be our main **categorical column** for grouping.


## 2. Quick Recap

- Data is usually loaded using `pd.read_csv()` or, as we just saw, directly from a library like `sklearn`
- Basic row/column selection uses `.loc[]`, `.iloc[]`, and boolean masks
- Today we go further: filtering with more complex conditions, sorting, and grouping — the operations you will use constantly during EDA.


## 3. Filtering Data — Single Condition


```python
df[df['wine_class'] == 'class_0']
```

```python
df[df['alcohol'] > 14]
```

```python
df[df['color_intensity'] >= 5]
```

## 4. Filtering Data — Multiple Conditions

Use `&` for AND, `|` for OR, and `~` for NOT. Each condition must be wrapped in parentheses.


```python
# AND condition
df[(df['wine_class'] == 'class_0') & (df['alcohol'] > 14)]
```

```python
# OR condition
df[(df['wine_class'] == 'class_1') | (df['wine_class'] == 'class_2')]
```

```python
# NOT condition
df[~(df['wine_class'] == 'class_0')]
```

```python
# Filter rows with alcohol > 13 and wine_class = class_0 or class_2
df[(df['alcohol'] > 13) & ((df['wine_class'] == 'class_0') | (df['wine_class'] == 'class_2'))]
```

## 5. Filtering Using `isin()` and `between()`


```python
df[df['wine_class'].isin(['class_0', 'class_2'])]
```

```python
df[df['alcohol'].between(12, 13)]
```

## 6. Filtering with String Conditions

Pandas provides a `.str` accessor for text-based filtering on object (string) columns.


```python
df[df['wine_class'].str.startswith('class_1')]
```

```python
df[df['wine_class'].str.contains('2')]
```

```python
df[df['wine_class'].str.len() == 7]
```

## 7. Filtering with `query()`

An alternative, more readable syntax for filtering, especially with multiple conditions.


```python
df.query('alcohol > 13 and wine_class == "class_0"')
```

```python
df.query('color_intensity >= 5 or hue > 1.0')
```

## 8. Sorting Data

Sorting is done using `.sort_values()`.


```python
df.sort_values('alcohol')                          # ascending by default
```

```python
df.sort_values('alcohol', ascending=False)         # descending
```

```python
# Sort by wine_class first, then by alcohol in ascending order
df.sort_values(['wine_class', 'alcohol'])          # sort by multiple columns
```

```python
# Sort by wine_class in ascending order and alcohol in descending order
df.sort_values(['wine_class', 'alcohol'], ascending=[True, False])   # mixed order per column
```

```python
# Sort the DataFrame by row index in ascending order
df.sort_index()                                     # sort by row index
```

### Sorting + Selecting Top/Bottom Records


```python
df.sort_values('proline', ascending=False).head(3)     # top 3 samples by proline content
```

```python
df.nlargest(3, 'proline')     # shortcut for top 3 by a column
```

```python
df.nsmallest(3, 'proline')    # shortcut for bottom 3 by a column
```

## 9. Grouping Data — Introduction

`groupby()` splits data into groups based on column values, then lets you apply an aggregate function to each group. This follows the **split → apply → combine** pattern.


```python
df.groupby('wine_class')['alcohol'].mean()
```

```python
df.groupby('wine_class')['proline'].max()
```

```python
df.groupby('wine_class')['color_intensity'].sum()
```

## 10. Grouping with Multiple Columns

Our dataset does not have a second natural categorical column, so let's create one — binning `alcohol` into levels — to demonstrate multi-column grouping.


```python
df.groupby(['wine_class', 'alcohol_level'])['proline'].mean()
```

```python
# Calculate the average color_intensity for each alcohol_level and wine_class
df.groupby(['alcohol_level', 'wine_class'])['color_intensity'].mean().reset_index()
```

**Tip:** `.reset_index()` converts the grouped result back into a regular DataFrame with a default numeric index, which is often easier to work with afterward.


## 11. Multiple Aggregations at Once

`.agg()` allows applying several aggregate functions in a single call.


```python
df.groupby('wine_class')['alcohol'].agg(['sum', 'mean', 'max', 'min', 'count'])
```

```python
df.groupby('wine_class').agg({
    'alcohol': 'mean',
    'proline': 'max',
    'color_intensity': 'mean',
    'hue': 'min'
})
```

## 13. `value_counts()` — A Simple Grouping Shortcut

`value_counts()` is a quick way to count how many times each unique value appears — effectively a groupby + count in one step.


```python
df['wine_class'].value_counts()
```

```python
df['alcohol_level'].value_counts(normalize=True)   # as proportions instead of counts
```

## 15. Putting It All Together — Mini Example

**Question:** Among wines with `color_intensity` greater than 4, which `wine_class` has the highest average `proline`, sorted from highest to lowest?


```python
result = (
    df[df['color_intensity'] > 4]
    .groupby('wine_class')['proline']
    .mean()
    .sort_values(ascending=False)
    .reset_index()
)
result
```

## Key Terms Glossary

| Term | Meaning |
|---|---|
| Boolean filtering | Selecting rows using a True/False condition |
| `&`, `\|`, `~` | AND, OR, NOT operators for combining conditions |
| `isin()` | Check membership in a list of values |
| `between()` | Check if values fall within a range |
| `.str` accessor | Access string methods on a text column |
| `query()` | Filter using a readable string expression |
| `sort_values()` | Sort rows by one or more column values |
| `sort_index()` | Sort rows by index |
| `nlargest()` / `nsmallest()` | Shortcut to get top/bottom N rows by a column |
| `groupby()` | Split data into groups based on column values |
| `agg()` | Apply one or more aggregate functions to grouped data |
| `value_counts()` | Count occurrences of each unique value |
| `pd.cut()` | Bin continuous numeric values into categorical ranges |
| `reset_index()` | Convert a grouped/indexed result back into a flat DataFrame |


## Common Mistakes to Avoid

- Forgetting parentheses around each condition when combining with `&` / `|` (e.g. `df[df['A'] > 5 & df['B'] < 10]` will error — needs `df[(df['A'] > 5) & (df['B'] < 10)]`)
- Using Python's `and`/`or` instead of `&`/`|` for combining conditions on Series
- Forgetting `ascending=False` and assuming `sort_values()` sorts descending by default
- Forgetting to select a column before aggregating (`df.groupby('wine_class').sum()` sums *all* numeric columns, which may not be intended)
- Not using `.reset_index()` after `groupby()` when a flat DataFrame is needed for further steps like merging or plotting
